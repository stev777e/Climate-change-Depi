## Handle Missing Values
Local Mean Imputation

```python
# Create a copy of the original dataset

df_sector_clean = df_sector.copy()


# Remove records with all annual values missing

df_sector_clean = df_sector_clean.dropna(
    subset=year_cols,
    how="all"
).reset_index(drop=True)


# Define local mean imputation function

def local_mean_imputation(row):

    original_values = row[year_cols].copy()
    filled_values = original_values.copy()

    # Calculate the overall row mean

    row_mean = original_values.mean()

    for i in range(len(year_cols)):

        # Skip non-missing values

        if pd.isna(original_values.iloc[i]):

            # Select 5 previous and 5 following years

            start = max(0, i - 5)
            end = min(len(year_cols), i + 6)

            # Get available neighboring values

            neighbors = pd.concat([
                original_values.iloc[start:i],
                original_values.iloc[i + 1:end]
            ]).dropna()

            # Apply local mean or overall row mean

            if not neighbors.empty:
                filled_values.iloc[i] = neighbors.mean()
            else:
                filled_values.iloc[i] = row_mean

    return filled_values


# Apply imputation to each record

df_sector_filled = df_sector_clean.copy()

df_sector_filled[year_cols] = df_sector_clean.apply(
    local_mean_imputation,
    axis=1
)


# Verify missing values

print(
    "Remaining Missing Values:",
    df_sector_filled[year_cols].isna().sum().sum()
)
```
